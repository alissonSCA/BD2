```python
import sqlite3
from pathlib import Path
import pandas as pd
```

## Passo 1: Garantir banco local
Este notebook usa o arquivo local `data/SQLite_databases_for_learning_data_science/iris.db`.
Se o arquivo nao existir, clone o repositorio de bancos dentro da pasta `data` antes de continuar.

```python
db_path = Path('data/SQLite_databases_for_learning_data_science/iris.db')

if not db_path.exists():
    raise FileNotFoundError(
        'Banco nao encontrado. Execute: mkdir -p data && git clone https://github.com/davidjamesknight/SQLite_databases_for_learning_data_science.git data/SQLite_databases_for_learning_data_science'
    )

print('Banco encontrado em', db_path)
```

```python
sql = '''
SELECT
    O.petal_length,
    O.petal_width,
    O.sepal_length,
    O.sepal_width,
    S.species
FROM Observation AS O
JOIN Species AS S ON S.species_id = O.species_id
'''

with sqlite3.connect(db_path) as conn:
    df = pd.read_sql_query(sql, conn)

print('Shape:', df.shape)
df.head()
```

```python
# Validacoes minimas do entregavel
assert not df.empty, 'DataFrame vazio'
assert {'petal_length', 'petal_width', 'sepal_length', 'sepal_width', 'species'}.issubset(df.columns)
print('Entregavel Semana 1 OK: conexao banco-pandas funcionando.')
```
